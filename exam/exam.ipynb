```python
#TASK 4

'''4. In a new Code block, write a function that takes as its parameter a list of strings. The function should return
True if one of the strings is coconut (in lower case, as shown), or False otherwise. Include some code
outside the function that shows that the function works correctly. Again, you may assume the "Happy Path"
so the list will always contain just strings.'''


def contains_coconut(words):
    return "coconut" in words

# Test the function
test_list_1 = ["apple", "banana", "coconut", "orange"]
test_list_2 = ["apple", "banana", "orange"]

print(contains_coconut(test_list_1))  
print(contains_coconut(test_list_2))  
```

```python
# TASK 5

''' 5.In a last new Code block, write a program that prompts the user to enter strings, one at a time. The input
ends when the user just presses "Enter" (so the string entered is empty). Then the program should display
the strings, in lower case, in alphabetical order'''

strings = []

while True:
    user_input = input("Enter a string (press Enter to finish): ")
    if user_input == "":
        break
    strings.append(user_input)

strings_lower = [s.lower() for s in strings]
strings_lower.sort() 

print("Strings in alphabetical order (lower case):")
for s in strings_lower:
    print(s)
```
